```python

# ============================================================
# MODULE 2: DATA PREPROCESSING
# Dataset: UCI Default of Credit Card Clients
# ============================================================

# STEP 1: IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

print("Libraries imported successfully.")


# STEP 2: LOAD THE DATASET

df = pd.read_csv("UCI_Credit_Card.csv")

# Remove any unwanted spaces in column names
df.columns = df.columns.str.strip()

print("\nDataset loaded successfully.")
print("Original dataset shape:", df.shape)


# STEP 3: DEFINE TARGET AND INPUT FEATURES

target = "default.payment.next.month"

# Check whether the target column exists
if target not in df.columns:
    raise ValueError(f"Target column '{target}' was not found.")

# Remove ID because it is only a record identifier
df_model = df.drop(columns=["ID"], errors="ignore")

# Separate input features (X) and target variable (y)
X = df_model.drop(columns=[target])
y = df_model[target]

print("\nInput features:", X.shape[1])
print("Target variable:", target)


# STEP 4: IDENTIFY CATEGORICAL AND NUMERICAL FEATURES

# Categorical features
categorical_features = [
    "SEX",
    "EDUCATION",
    "MARRIAGE"
]

# Verify that all categorical columns exist
missing_columns = [
    col for col in categorical_features
    if col not in X.columns
]

if missing_columns:
    raise ValueError(f"Missing categorical columns: {missing_columns}")

# All remaining features are treated as numerical
numerical_features = [
    col for col in X.columns
    if col not in categorical_features
]

print("\nNumber of categorical features:", len(categorical_features))
print("Number of numerical features:", len(numerical_features))


# STEP 5: SPLIT DATA INTO TRAINING AND TESTING SETS

# 80% data for training and 20% for testing
# Stratification maintains the original target-class proportions

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining and testing split completed.")
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)


# STEP 6: CREATE THE STANDARD SCALER

# StandardScaler standardises numerical features
numeric_transformer = StandardScaler()


# STEP 7: CREATE THE ONE HOT ENCODER

# Handle different Scikit-learn versions
try:
    categorical_transformer = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    categorical_transformer = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )


# STEP 8: BUILD THE COLUMN TRANSFORMER

# Apply StandardScaler to numerical features
# Apply OneHotEncoder to categorical features

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("\nColumnTransformer created successfully.")


# STEP 9: FIT AND TRANSFORM THE TRAINING DATA

# Learn scaling parameters and category encodings from training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply the same learned transformations to test data
X_test_processed = preprocessor.transform(X_test)

print("\nData preprocessing completed successfully.")


# STEP 10: VALIDATE THE PROCESSED DATA

print("\n========== PREPROCESSING RESULTS ==========")

print("\nOriginal training data:", X_train.shape)
print("Processed training data:", X_train_processed.shape)

print("\nOriginal testing data:", X_test.shape)
print("Processed testing data:", X_test_processed.shape)

# Check for missing and infinite values
print("\nMissing values in training data:",
      np.isnan(X_train_processed).sum())

print("Missing values in testing data:",
      np.isnan(X_test_processed).sum())

print("\nInfinite values in training data:",
      np.isinf(X_train_processed).sum())

print("Infinite values in testing data:",
      np.isinf(X_test_processed).sum())

# Display processed feature names
processed_feature_names = preprocessor.get_feature_names_out()

print("\nTotal processed features:", len(processed_feature_names))
print("\nSample processed feature names:")
print(processed_feature_names[:10])

print("\n========== MODULE 2 COMPLETED ==========")
```

```python
# ============================================
# MODULE 4: MODEL EVALUATION
# SECTION 4.1: IMPORT LIBRARIES
# ============================================

# Model evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

# Machine learning pipeline tools
from sklearn.pipeline import Pipeline
from sklearn.base import clone

# Classification algorithms
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import LinearSVC

# Preprocessing tools
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All model evaluation libraries imported successfully.")
```

```python
# ============================================
# SECTION 4.2: CREATE PREPROCESSING TRANSFORMER
# ============================================

# Identify categorical features
categorical_features = [
    "SEX",
    "EDUCATION",
    "MARRIAGE"
]

# Identify numerical features
numerical_features = [
    column for column in X_train.columns
    if column not in categorical_features
]

# Standardise numerical features
numeric_transformer = StandardScaler()

# Encode categorical features
try:
    categorical_transformer = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    categorical_transformer = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

# Combine both operations
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Numerical features:", len(numerical_features))
print("Categorical features:", len(categorical_features))
print("Preprocessor created successfully.")
```

```python
# ============================================
# SECTION 4.3: DEFINE CLASSIFICATION MODELS
# ============================================

classifiers = {

    # Model 1: Logistic Regression
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    # Model 2: Decision Tree
    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    # Model 3: K-Nearest Neighbours
    "K-Nearest Neighbours": KNeighborsClassifier(
        n_neighbors=5,
        n_jobs=-1
    ),

    # Model 4: Gaussian Naive Bayes
    "Gaussian Naive Bayes": GaussianNB(),

    # Model 5: Linear Support Vector Machine
    "Support Vector Machine": LinearSVC(
        C=1.0,
        max_iter=10000,
        random_state=42
    )
}

print("Five classification algorithms defined successfully.")

for model_name in classifiers:
    print("-", model_name)
```

```python
# ============================================
# SECTION 4.4: CREATE MODEL PIPELINES
# ============================================

model_pipelines = {}

for model_name, classifier in classifiers.items():

    model_pipelines[model_name] = Pipeline(
        steps=[
            ("preprocessor", clone(preprocessor)),
            ("classifier", classifier)
        ]
    )

print("Five independent pipelines created successfully.")

for model_name in model_pipelines:
    print("-", model_name)
```

```python
# ============================================
# SECTION 4.5: TRAIN MODELS AND GENERATE PREDICTIONS
# ============================================

# Dictionary to store trained models
trained_models = {}

# Store predicted classes for each model
predictions = {}

# Store continuous scores for AUC calculation
auc_scores = {}

# Train and evaluate predictions
for model_name, pipeline in model_pipelines.items():

    print(f"\nTraining: {model_name}")

    # Train the complete pipeline
    pipeline.fit(X_train, y_train)

    # Store the trained pipeline
    trained_models[model_name] = pipeline

    # Predict default status on test data
    y_pred = pipeline.predict(X_test)

    # Save predictions
    predictions[model_name] = y_pred

    # Obtain scores for calculating AUC
    if hasattr(pipeline, "predict_proba"):

        # Use predicted probability of class 1
        positive_index = list(pipeline.classes_).index(1)

        y_score = pipeline.predict_proba(X_test)[:, positive_index]

    else:

        # Use decision scores for LinearSVC
        y_score = pipeline.decision_function(X_test)

        # Ensure higher scores represent class 1
        positive_index = list(pipeline.classes_).index(1)

        if positive_index == 0:
            y_score = -y_score

    auc_scores[model_name] = y_score

    print(f"{model_name} training and prediction completed.")

print("\nAll five models trained successfully.")
```

```python
# ============================================
# SECTION 4.6: CALCULATE EVALUATION METRICS
# ============================================

# Store evaluation results
results = []

for model_name in trained_models:

    # Actual target values
    y_actual = y_test

    # Predicted target values
    y_pred = predictions[model_name]

    # Continuous scores for AUC
    y_score = auc_scores[model_name]

    # Calculate six evaluation metrics
    accuracy = accuracy_score(y_actual, y_pred)

    auc = roc_auc_score(y_actual, y_score)

    precision = precision_score(
        y_actual,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    recall = recall_score(
        y_actual,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    f1 = f1_score(
        y_actual,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    mcc = matthews_corrcoef(y_actual, y_pred)

    # Store results
    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "AUC": auc,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "MCC": mcc
    })


# Convert results into a DataFrame
results_df = pd.DataFrame(results)

# Set model as index
results_df = results_df.set_index("Model")

# Display rounded evaluation results
print("MODEL PERFORMANCE COMPARISON")
display(results_df.round(4))
```

```python
# ============================================
# SECTION 4.7: CONFUSION MATRICES
# ============================================

for model_name, y_pred in predictions.items():

    # Create a separate confusion matrix for each model
    fig, ax = plt.subplots(figsize=(6, 5))

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        y_pred,
        display_labels=["No Default (0)", "Default (1)"],
        cmap="Blues",
        values_format="d",
        ax=ax
    )

    ax.set_title(f"Confusion Matrix - {model_name}")

    plt.tight_layout()
    plt.show()
```

```python
# ============================================
# SECTION 4.8: CLASSIFICATION REPORTS
# ============================================

for model_name, y_pred in predictions.items():

    print("\n" + "=" * 60)
    print(f"CLASSIFICATION REPORT: {model_name}")
    print("=" * 60)

    print(
        classification_report(
            y_test,
            y_pred,
            target_names=["No Default", "Default"],
            zero_division=0
        )
    )
```
